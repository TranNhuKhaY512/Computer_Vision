## 6.Thực hành với Cân bằng sáng - equalizeHist, để cân bằng sáng, trước hết ta chuyển ảnh trắng đen (nếu có) sang ảnh màu.

```python
import cv2
import numpy as np
import matplotlib.pyplot as TranNhuKhaY

image = cv2.imread("cat.jpg")
TranNhuKhaY.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
TranNhuKhaY.axis("off")
TranNhuKhaY.title("Original")
TranNhuKhaY.show()
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

## 1.Sử dụng Backward, forward, central gradient để lọc ảnh. 

```python
# Forward gradient (dx)
kx_f = np.array([[0, -1, 1]], dtype=np.float32)

# Backward gradient (dx)
kx_b = np.array([[-1, 1, 0]], dtype=np.float32)

# Central gradient (dx)
kx_c = np.array([[-1, 0, 1]], dtype=np.float32) / 2

forward  = cv2.filter2D(gray, -1, kx_f)
backward = cv2.filter2D(gray, -1, kx_b)
central  = cv2.filter2D(gray, -1, kx_c)

TranNhuKhaY.figure(figsize=(12,4))
TranNhuKhaY.subplot(1,3,1); TranNhuKhaY.imshow(forward, cmap="gray");  TranNhuKhaY.title("Forward");  TranNhuKhaY.axis("off")
TranNhuKhaY.subplot(1,3,2); TranNhuKhaY.imshow(backward,cmap="gray");  TranNhuKhaY.title("Backward"); TranNhuKhaY.axis("off")
TranNhuKhaY.subplot(1,3,3); TranNhuKhaY.imshow(central, cmap="gray");  TranNhuKhaY.title("Central");  TranNhuKhaY.axis("off")
TranNhuKhaY.show()
```

## 2. Finite difference filter để lọc ảnh nhiễu

```python
k_fd_x = np.array([[1, -1]], dtype=np.float32)
k_fd_y = np.array([[1], [-1]], dtype=np.float32)

fd_x = cv2.filter2D(gray, -1, k_fd_x)
fd_y = cv2.filter2D(gray, -1, k_fd_y)

fd = cv2.addWeighted(fd_x, 0.5, fd_y, 0.5, 0)

TranNhuKhaY.figure(figsize=(12,4))
TranNhuKhaY.subplot(1,3,1); TranNhuKhaY.imshow(fd_x, cmap="gray"); TranNhuKhaY.title("Finite diff X"); TranNhuKhaY.axis("off")
TranNhuKhaY.subplot(1,3,2); TranNhuKhaY.imshow(fd_y, cmap="gray"); TranNhuKhaY.title("Finite diff Y"); TranNhuKhaY.axis("off")
TranNhuKhaY.subplot(1,3,3); TranNhuKhaY.imshow(fd,   cmap="gray"); TranNhuKhaY.title("Combine");       TranNhuKhaY.axis("off")
TranNhuKhaY.show()
```

## 3. Gaussian filter để lọc ảnh

```python
gauss = cv2.GaussianBlur(gray, (5, 5), 0)

TranNhuKhaY.figure(figsize=(10,4))
TranNhuKhaY.subplot(1,2,1); TranNhuKhaY.imshow(gray,  cmap="gray"); TranNhuKhaY.title("Gray");     TranNhuKhaY.axis("off")
TranNhuKhaY.subplot(1,2,2); TranNhuKhaY.imshow(gauss, cmap="gray"); TranNhuKhaY.title("Gaussian"); TranNhuKhaY.axis("off")
TranNhuKhaY.show()
```

## 4. Sobel detector để xác định biên

```python
sx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sy = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

mag = cv2.magnitude(sx, sy)
sobel_edge = cv2.convertScaleAbs(mag)

TranNhuKhaY.figure(figsize=(12,4))
TranNhuKhaY.subplot(1,3,1); TranNhuKhaY.imshow(cv2.convertScaleAbs(sx), cmap="gray"); TranNhuKhaY.title("Sobel X"); TranNhuKhaY.axis("off")
TranNhuKhaY.subplot(1,3,2); TranNhuKhaY.imshow(cv2.convertScaleAbs(sy), cmap="gray"); TranNhuKhaY.title("Sobel Y"); TranNhuKhaY.axis("off")
TranNhuKhaY.subplot(1,3,3); TranNhuKhaY.imshow(sobel_edge, cmap="gray");              TranNhuKhaY.title("Sobel Mag"); TranNhuKhaY.axis("off")
TranNhuKhaY.show()



```

## 5. Canny edge detector để xác định cạnh

```python
edges = cv2.Canny(gray, 100, 200)

TranNhuKhaY.figure(figsize=(10,4))
TranNhuKhaY.subplot(1,2,1); TranNhuKhaY.imshow(gray,  cmap="gray"); TranNhuKhaY.title("Gray");  TranNhuKhaY.axis("off")
TranNhuKhaY.subplot(1,2,2); TranNhuKhaY.imshow(edges, cmap="gray"); TranNhuKhaY.title("Canny"); TranNhuKhaY.axis("off")
TranNhuKhaY.show()
```

## 6. Hough transform để xác định đường thẳng

```python
edges = cv2.Canny(gray, 100, 200)

lines = cv2.HoughLinesP(
    edges,
    rho=1,
    theta=np.pi/180,
    threshold=100,
    minLineLength=50,
    maxLineGap=10
)

img_lines = image.copy()
if lines is not None:
    for x1, y1, x2, y2 in lines[:, 0]:
        cv2.line(img_lines, (x1, y1), (x2, y2), (0, 0, 255), 2)

TranNhuKhaY.imshow(cv2.cvtColor(img_lines, cv2.COLOR_BGR2RGB))
TranNhuKhaY.axis("off")
TranNhuKhaY.title("Hough Lines")
TranNhuKhaY.show()
```
